```python
def dfs(graph, start, goal, path=None, visited=None):
    """Finds a path from start to goal in a graph using Depth-First Search (DFS)."""
    if path is None:
        path = [start]
    if visited is None:
        visited = {start}

    if start == goal:
        return path

    for neighbor in graph.get(start, []):
        if neighbor not in visited:
            visited.add(neighbor)
            result = dfs(graph, neighbor, goal, path + [neighbor], visited)
            if result:
                return result
    return None

# Sample city map dataset to make the code runnable
city_map = {
    "Tehran": ["Karaj", "Qom", "Hamedan"],
    "Karaj": ["Qazvin"],
    "Qazvin": ["Hamedan"],
    "Qom": ["Isfahan"],
    "Hamedan": []
}

# Run the DFS algorithm
print(dfs(city_map, "Tehran", "Hamedan"))
```
